```python
model_id = "meta-llama/Meta-Llama-3.1-8B"
pipeline = transformers.pipeline(
    "text-generation",
    model=model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)
```

```python
for _ in range(5):
    print(pipeline("Hello I am ", max_length=400))
```
